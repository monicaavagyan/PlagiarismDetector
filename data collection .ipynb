```python
import aiohttp
import asyncio
import pandas as pd
import random
import time
import schedule

# CORE API Key (Replace with your actual API key)
CORE_API_KEY = "4VxnqIdpTFWyU95ojutDAS6aBROvc708"

# CORE API Endpoint
CORE_API_URL = "https://api.core.ac.uk/v3/search/works"

# File to store collected papers
CSV_FILE = "core_papers.csv"

# List of research topics to fetch papers from various fields
TOPICS = [
    "Artificial Intelligence", "Quantum Computing", "Biotechnology", 
    "Blockchain", "Climate Change", "Cybersecurity", "Economics",
    "Neuroscience", "Data Science", "Robotics", "Space Exploration",
    "Renewable Energy", "Bioinformatics", "Linguistics", "Finance"
]

# Function to fetch papers from CORE API
async def fetch_core_papers(session, topic, max_results=20):
    headers = {"Authorization": f"Bearer {CORE_API_KEY}"}
    params = {
        "q": topic,
        "limit": max_results
    }

    async with session.get(CORE_API_URL, headers=headers, params=params) as response:
        if response.status == 200:
            papers = await response.json()
            paper_list = []

            for paper in papers.get("results", []):
                paper_list.append({
                    "Title": paper.get("title", "N/A"),
                    "Authors": ", ".join([a.get("name", "Unknown") for a in paper.get("authors", [])]),
                    "Published Date": paper.get("year", "N/A"),
                    "Abstract": paper.get("abstract", "No abstract available"),
                    "Topic": topic,
                    "DOI": paper.get("doi", "N/A"),
                    "Full Text Link": paper.get("fullTextLink", "N/A"),
                })
            return paper_list
        else:
            print(f"Error {response.status}: {await response.text()}")
            return []

# Function to save collected data
def save_to_csv(papers):
    df = pd.DataFrame(papers)

    # Append new data while avoiding duplicates
    try:
        df_existing = pd.read_csv(CSV_FILE)
        df = pd.concat([df_existing, df]).drop_duplicates(subset=["DOI"]).reset_index(drop=True)
    except FileNotFoundError:
        pass  # First-time run, no file yet

    df.to_csv(CSV_FILE, index=False)
    print(f"Saved {len(papers)} new papers to {CSV_FILE}")

# Function to automate collection from multiple topics
async def automate_collection():
    print("Fetching new research papers from CORE API...")
    selected_topic = random.choice(TOPICS)  # Select random topic each time
    async with aiohttp.ClientSession() as session:
        papers = await fetch_core_papers(session, topic=selected_topic, max_results=50)

    if papers:
        save_to_csv(papers)
    else:
        print("No new papers found.")

# Schedule data collection every 24 hours
schedule.every().day.at("12:00").do(lambda: asyncio.run(automate_collection()))

print("Automation started... Press Ctrl+C to stop.")
while True:
    schedule.run_pending()
    time.sleep(1)
```

###  Explanation 
✅ Randomly selects a research topic <br>
✅ Fetches 50 papers per day from CORE API<br>
✅ Stores metadata & full-text links for analysis<br>
✅ Avoids duplicate papers using DOI<br>
✅ Runs in the background using schedule
