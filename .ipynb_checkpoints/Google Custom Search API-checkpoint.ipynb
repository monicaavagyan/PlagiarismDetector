### Այս կոդը նախատեսված է տեքստի յուրահատկությունը ստուգելու համար՝ օգտագործելով Google Custom Search API: Այն ստուգում է, թե արդյոք տրված տեքստը կա ինտերնետում, և եթե այո, ապա տալիս է այն աղբյուրները, որտեղ տեքստը գտնվել է: 

## For JSON format

```python
import requests

def check_uniqueness(api_key, cx, query):
    url = 'https://www.googleapis.com/customsearch/v1'
    params = {
        'key': api_key,
        'cx': cx,
        'q': query
    }
    response = requests.get(url, params=params)
    results = response.json()

    return results

def analyze_results(results):
    if 'items' in results:
        print("The text is likely to be copied. Here are some sources where it was found:")
        for item in results['items']:
            print(f"Title: {item['title']}")
            print(f"Link: {item['link']}")
            print()
    else:
        print("The text appears to be unique.")

if __name__ == "__main__":
    API_KEY = 'Your API key' # Replace with your API key
    CX = 'Your search engine ID'  # Replace with your Custom Search Engine ID
    query = input("Enter the text to check for uniqueness: ")

    results = check_uniqueness(API_KEY, CX, query)
    analyze_results(results)
```

## Works with sentences 

### Այս կոդը կարելի է օգտագործել  գրագողության հակազդման համակարգում՝ տեքստի յուրահատկությունը(uniqnes) ստուգելու համար: 

```python
import requests
import logging
import urllib.parse

# Configure logging
logging.basicConfig(level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')

def check_uniqueness(api_key, cx, query):
    url = 'https://www.googleapis.com/customsearch/v1'
    params = {
        'key': api_key,
        'cx': cx,
        'q': query
    }
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()  # Raise an exception for HTTP errors
        results = response.json()
        return results
    except requests.exceptions.RequestException as e:
        logging.error(f"Request failed: {e}")
        return None
    except requests.exceptions.JSONDecodeError:
        logging.error("Error: Unable to decode JSON response. Please check your API key and Custom Search Engine ID.")
        return None

def analyze_results(results):
    if results is None:
        print("No results to analyze.")
        return

    if 'items' in results:
        print("The text is likely to be copied. Here are some sources where it was found:")
        for item in results['items']:
            print(f"Title: {item['title']}")
            print(f"Link: {item['link']}")
            print()
    else:
        print("The text appears to be unique.")

if __name__ == "__main__":
    API_KEY = 'Your API key' # Replace with your API key
    CX = 'Your search engine ID'  # Replace with your Custom Search Engine ID
    query = input("Enter the text to check for uniqueness: ")

    results = check_uniqueness(API_KEY, CX, query)
    analyze_results(results)
```

### Searching an image via text

```python
import requests
import logging

# Configure logging
logging.basicConfig(level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')

def check_uniqueness(api_key, cx, query):
    url = 'https://www.googleapis.com/customsearch/v1'
    params = {
        'key': api_key,
        'cx': cx,
        'q': query,
        'searchType': 'image'  # Add this parameter for image search
    }
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()  # Raise an exception for HTTP errors
        results = response.json()
        return results
    except requests.exceptions.RequestException as e:
        logging.error(f"Request failed: {e}")
        return None
    except requests.exceptions.JSONDecodeError:
        logging.error("Error: Unable to decode JSON response. Please check your API key and Custom Search Engine ID.")
        return None

def analyze_results(results):
    if results is None:
        print("No results to analyze.")
        return

    if 'items' in results:
        print("Matching images found. Here are the results:")
        for item in results['items']:
            print(f"Title: {item['title']}")
            print(f"Image URL: {item['link']}")
            print(f"Image Thumbnail: {item['image']['thumbnailLink']}")
            print()
    else:
        print("No matching images found.")

if __name__ == "__main__":
    API_KEY = 'Your API key' # Replace with your API key
    CX = 'Your search engine ID'  # Replace with your Custom Search Engine ID
    query = input("Enter the text to search for images: ")

    results = check_uniqueness(API_KEY, CX, query)
    analyze_results(results)
```

```python
import os
from tkinter import Tk, filedialog
from google.cloud import vision
from google.cloud.vision_v1 import types
from google.api_core.exceptions import GoogleAPICallError, InvalidArgument

# Set up Google Cloud credentials
os.environ['GOOGLE_APPLICATION_CREDENTIALS'] =  r'C:\Users\ADMIN\Downloads\vertexprojectfortranslation-c44af0486ef4.json'  # Replace with your service account key file

def search_similar_images(image_path):
    try:
        # Initialize the Vision API client
        client = vision.ImageAnnotatorClient()

        # Load the image
        with open(image_path, 'rb') as image_file:
            content = image_file.read()

        image = types.Image(content=content)

        # Perform web detection to find similar images
        response = client.web_detection(image=image)
        annotations = response.web_detection

        if annotations.pages_with_matching_images:
            print(f"Found {len(annotations.pages_with_matching_images)} similar images:")
            for page in annotations.pages_with_matching_images:
                print(f"URL: {page.url}")
        else:
            print("No similar images found.")
    except (GoogleAPICallError, InvalidArgument) as e:
        print(f"Error occurred while searching for similar images: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

def upload_image():
    # Create a Tkinter root window (it will be hidden)
    root = Tk()
    root.withdraw()  # Hide the root window

    # Open a file dialog to select an image
    image_path = filedialog.askopenfilename(
        title="Select an image",
        filetypes=[("Image files", "*.jpg *.jpeg *.png *.bmp *.gif")]
    )

    if image_path:
        print(f"Selected image: {image_path}")
        search_similar_images(image_path)
    else:
        print("No image selected.")

if __name__ == "__main__":
    upload_image()
```
